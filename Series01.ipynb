```python
import numpy as np
import matplotlib.pyplot as plt

# Charger les données
ecg_raw = np.loadtxt('data/ECGu.txt')
print("Shape of ecg_raw:", ecg_raw.shape)  # (6500, 3)

# Paramètres
sampling_frequency = 1000  # Hz
T = 1 / sampling_frequency  # période d'échantillonnage
N = ecg_raw.shape[0]        # nombre d’échantillons = 6500
start = 0
stop = (N - 1) * T
t = np.linspace(start, stop, N)

print("Shape of t:", t.shape)
print("First 10 values of t:", t[:10])

# Tracer la 1ère dérivation (colonne 0)
plt.plot(t, ecg_raw[:, 0], label="Lead I")
plt.title("ECG Signal - Lead I")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude (ADU)")
plt.grid(True)
plt.legend()
plt.show()

# Lead II
plt.plot(t, ecg_raw[:, 1], label="Lead II")
plt.title("ECG Signal - Lead II")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude (ADU)")
plt.grid(True)
plt.legend()
plt.show()

# Lead III
plt.plot(t, ecg_raw[:, 2], label="Lead III")
plt.title("ECG Signal - Lead III")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude (ADU)")
plt.grid(True)
plt.legend()
plt.show()

```

```python
fig, axs = plt.subplots(3, 1, figsize=(12, 6), sharex=True)

leads = ["Lead I", "Lead II", "Lead III"]

for i in range(3):
    axs[i].plot(t, ecg_raw[:, i], color="black", linewidth=1)
    axs[i].set_ylabel(leads[i], rotation=0, labelpad=20, fontsize=12, va="center")
    axs[i].grid(True)

axs[-1].set_xlabel("Time (s)")
plt.tight_layout()
plt.show()
```

```python
# rescale the ecg signal knowing that 1 unit = 1024 microV
ecg = ecg_raw * 1024 / 1000 * 1e-6  # en mV

# plot the ecg as a function of time
plt.plot(t, ecg)
plt.xlabel("Time (s)")        # <-- mettre entre guillemets
plt.ylabel("Amplitude (mV)")  # <-- mettre entre guillemets
plt.grid(True)

# plt.title("ECG signal")
plt.show()
```
